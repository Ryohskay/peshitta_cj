```python
from pathlib import Path
```

```python
for book in Fs("book@en").items():
    if book[1] in target_books.keys():
        print(f"\nBook: {book[1]}\n")
        save_file = Path(f"./out/ot_etcbc_{book[1]}_roots_results.csv")
        
        chapters = L.d(book[0], otype="chapter")
        results = "Verse Reference,Lemmatised Transliteration,Plain Transliteration,Original Syriac Text\n"

        for chapter in chapters:
            for verse in L.d(chapter, otype="verse"):
                verse_ref = f"Chapter {F.chapter.v(chapter)} Verse {F.verse.v(verse)}"
                # Get all words in this verse
                words = L.d(verse, otype="word")
                # Lemmatised text of this verse, in transliterated form
                lemmata = []
                lemmata.extend([lemma for w_id in words for lemma in lemmatise(w_id)])
                verse_lemmata = ' '.join(lemmata)
                # Transliteration of this verse as a list
                words_translit = [F.word_etcbc.v(w_id) for w_id in words]
                verse_text = ' '.join(words_translit)
                # This verse in Syriac alphabet
                words_src = [F.word.v(w_id) for w_id in words]
                verse_src = ' '.join(words_src)
                print(f"{verse_ref}\n\t{verse_lemmata}\t{verse_text}\n\t{verse_src}")
                results += f"{verse_ref},{verse_lemmata},{verse_text},{verse_src}\n"

        save_file.write_text(results)

        # print(f"{book[1]}:\n\t{verse_txt}")
```
