```python
from tf_utils import get_verses
from pathlib import Path
import json
```

```python
ot_train_books = {
    "Genesis": list(range(1,51)),
    "Exodus": list(range(1,22))
}

ot_test_books = {
    "Deuteronomy": list(range(1,21))
}
```

```python
ot_train_book_verses = get_verses(ot_train_books)
```

```python
ot_test_book_verses = get_verses(ot_test_books)
```

```python
nt_train_books = {
    "Matthew": list(range(1,31)),
    "Mark": list(range(1,17)),
    "Luke": list(range(1,24)),
    "John": list(range(1,22)),
}

nt_test_books = {
    "Acts": list(range(1,29))
}
```

```python
nt_train_book_verses = get_verses(nt_train_books, target_fabric="etcbc/syrnt", ver="0.1")
```

```python
nt_test_book_verses = get_verses(nt_test_books, target_fabric="etcbc/syrnt", ver="0.1")
```

```python
ot_train_verses = [vrs for verses in ot_train_book_verses.values() for vrs in verses]
nt_train_verses = [vrs for verses in nt_train_book_verses.values() for vrs in verses]

train_verse_txts = ot_train_verses.copy()
train_verse_txts.extend(nt_train_verses)

print(train_verse_txts[0])
```

```python
ot_train_labels = [0 for i in range(len(ot_train_verses))]
nt_train_labels = [1 for i in range(len(nt_train_verses))]

train_labels = ot_train_labels.copy()
train_labels.extend(nt_train_labels)
```

```python
ot_test_verses = [vrs for verses in ot_test_book_verses.values() for vrs in verses]
nt_test_verses = [vrs for verses in nt_test_book_verses.values() for vrs in verses]

test_verse_txts = ot_test_verses.copy()
test_verse_txts.extend(nt_test_verses)

print(test_verse_txts[0])
```

```python
ot_test_labels = [0 for i in range(len(ot_test_verses))]
nt_test_labels = [1 for i in range(len(nt_test_verses))]

test_labels = ot_test_labels.copy()
test_labels.extend(nt_test_labels)
```

```python
def generate_verses(data: list, labels: list, form: str="translit"):
    # assert(len(data) == len(labels))
    if form == "translit":
        format_idx = 1 
    elif form == "syriac":
        format_idx = 2
    else:
        raise ValueError(f"form argument must be 'translit' or 'syriac', but got {form}")
        
    for i in range(len(data)):
        yield {"label": labels[i], "text": ' '.join(data[i][format_idx])}
```

```python
def save_as_json(fname: str, data: list, labels: list):
    """Save the given data in json compatible with Huggingface datasets."""
    verses = []
    
    for verse_jso in generate_verses(data, labels):
        verses.append(verse_jso)
    
    with Path(fname).open("w") as fp:
        json.dump(verses, fp)
    
    print(fname)
```

```python
save_as_json("./out/ot_train.json", ot_train_verses, ot_train_labels)
save_as_json("./out/nt_train.json", nt_train_verses, nt_train_labels)
save_as_json("./out/ot_test.json", ot_test_verses, ot_test_labels)
save_as_json("./out/nt_test.json", nt_test_verses, nt_test_labels)
```
